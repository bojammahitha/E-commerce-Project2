````python
# ============================================================

# E-COMMERCE AI DATA ANALYST

# GOOGLE COLAB - SINGLE CELL

# GEMINI + SQLITE + LANGCHAIN

# ============================================================

!pip -q install -U langchain-google-genai faker pandas gTTS

# ============================================================

# IMPORTS

# ============================================================

import os
import re
import sqlite3
import random
import pandas as pd

from faker import Faker
from getpass import getpass
from datetime import datetime, timedelta

from langchain_google_genai import ChatGoogleGenerativeAI
from gtts import gTTS
from IPython.display import Audio, display

# ============================================================

# GEMINI API KEY

# ============================================================

print("=" * 70)
print("GEMINI API KEY")
print("=" * 70)

GEMINI_API_KEY = getpass("Paste your Gemini API key: ")

os.environ["GOOGLE_API_KEY"] = GEMINI_API_KEY
os.environ["GEMINI_API_KEY"] = GEMINI_API_KEY

print("Gemini API key loaded successfully.")

# ============================================================

# CREATE DATABASE

# ============================================================

DB_PATH = "/content/ecommerce.db"

if os.path.exists(DB_PATH):
    os.remove(DB_PATH)

conn = sqlite3.connect(DB_PATH)

cursor = conn.cursor()

print("\nCreating SQLite database...")

# ============================================================

# CREATE TABLES

# ============================================================

cursor.execute("""
CREATE TABLE customers (
    id INTEGER PRIMARY KEY,
    name TEXT,
    email TEXT,
    city TEXT,
    state TEXT,
    created_at TEXT
)
""")

cursor.execute("""
CREATE TABLE products (
    id INTEGER PRIMARY KEY,
    name TEXT,
    category TEXT,
    price REAL,
    cost REAL,
    stock INTEGER
)
""")

cursor.execute("""
CREATE TABLE orders (
    id INTEGER PRIMARY KEY,
    customer_id INTEGER,
    order_date TEXT,
    status TEXT,
    total_amount REAL
)
""")

cursor.execute("""
CREATE TABLE order_items (
    id INTEGER PRIMARY KEY,
    order_id INTEGER,
    product_id INTEGER,
    quantity INTEGER,
    unit_price REAL
)
""")

cursor.execute("""
CREATE TABLE reviews (
    id INTEGER PRIMARY KEY,
    product_id INTEGER,
    customer_id INTEGER,
    rating INTEGER,
    review_text TEXT,
    review_date TEXT
)
""")

conn.commit()

print("Tables created successfully.")

# ============================================================

# GENERATE DATA

# ============================================================

random.seed(42)
Faker.seed(42)

fake = Faker("en_IN")

# ============================================================

# CUSTOMERS

# ============================================================

cities = [
    ("Bengaluru", "Karnataka"),
    ("Mysuru", "Karnataka"),
    ("Mumbai", "Maharashtra"),
    ("Pune", "Maharashtra"),
    ("Delhi", "Delhi"),
    ("Hyderabad", "Telangana"),
    ("Chennai", "Tamil Nadu"),
    ("Kochi", "Kerala"),
    ("Ahmedabad", "Gujarat"),
    ("Jaipur", "Rajasthan")
]

customers = []

for i in range(1, 501):

    city, state = random.choice(cities)

    created_at = (
        datetime.now()
        - timedelta(days=random.randint(30, 1000))
    ).strftime("%Y-%m-%d")

    customers.append(
        (
            i,
            fake.name(),
            fake.email(),
            city,
            state,
            created_at
        )
    )

cursor.executemany(
    """
INSERT INTO customers
(id, name, email, city, state, created_at)
VALUES (?, ?, ?, ?, ?, ?)
""",
    customers
)

# ============================================================

# PRODUCTS

# ============================================================

catalog = [

    ("Laptop Pro 14", "Electronics", 74999, 60000),
    ("Laptop Air 13", "Electronics", 64999, 51000),
    ("Gaming Laptop X", "Electronics", 89999, 72000),
    ("Wireless Mouse", "Electronics", 1499, 800),
    ("Mechanical Keyboard", "Electronics", 3499, 2100),
    ("Noise Cancelling Headphones", "Electronics", 7999, 5200),
    ("Bluetooth Speaker", "Electronics", 3999, 2500),
    ("Smart Watch", "Electronics", 5999, 3800),
    ("USB-C Hub", "Electronics", 2499, 1400),
    ("Power Bank", "Electronics", 1999, 1100),

    ("Running Shoes", "Fashion", 4999, 2800),
    ("Casual Sneakers", "Fashion", 3999, 2300),
    ("Formal Shoes", "Fashion", 5499, 3200),
    ("Denim Jacket", "Fashion", 3499, 2000),
    ("Cotton T-Shirt", "Fashion", 999, 450),
    ("Hoodie", "Fashion", 1999, 1000),

    ("Coffee Maker", "Home", 6999, 4300),
    ("Air Fryer", "Home", 8999, 5800),
    ("Mixer Grinder", "Home", 5499, 3400),
    ("Electric Kettle", "Home", 1999, 1100),
    ("Vacuum Cleaner", "Home", 11999, 7800),
    ("Desk Lamp", "Home", 1499, 700),

    ("Programming in Python", "Books", 799, 400),
    ("Data Science Handbook", "Books", 1299, 700),
    ("AI Engineering", "Books", 999, 550),
    ("Clean Code", "Books", 699, 350)

]

products = []

for i, item in enumerate(catalog, start=1):

    name = item[0]
    category = item[1]
    price = item[2]
    cost = item[3]

    stock = random.randint(5, 250)

    products.append(
        (
            i,
            name,
            category,
            price,
            cost,
            stock
        )
    )

cursor.executemany(
    """
INSERT INTO products
(id, name, category, price, cost, stock)
VALUES (?, ?, ?, ?, ?, ?)
""",
    products
)

# ============================================================

# ORDERS

# ============================================================

orders = []
order_items = []

product_ids = [p[0] for p in products]

product_prices = {
    p[0]: p[3]
    for p in products
}

item_id = 1

for order_id in range(1, 2501):

    customer_id = random.randint(1, 500)

    order_date = (
        datetime.now()
        - timedelta(days=random.randint(0, 365))
    ).strftime("%Y-%m-%d")

    status = random.choices(
        [
            "completed",
            "cancelled",
            "returned",
            "pending"
        ],
        weights=[80, 5, 10, 5],
        k=1
    )[0]

    selected_products = random.sample(
        product_ids,
        random.randint(1, 4)
    )

    total = 0

    for product_id in selected_products:

        quantity = random.randint(1, 3)

        price = product_prices[product_id]

        discount = random.choice(
            [0, 0, 0, 0.05, 0.10, 0.15]
        )

        selling_price = round(
            price * (1 - discount),
            2
        )

        total = total + (
            quantity * selling_price
        )

        order_items.append(
            (
                item_id,
                order_id,
                product_id,
                quantity,
                selling_price
            )
        )

        item_id = item_id + 1

    orders.append(
        (
            order_id,
            customer_id,
            order_date,
            status,
            round(total, 2)
        )
    )

cursor.executemany(
    """
INSERT INTO orders
(id, customer_id, order_date, status, total_amount)
VALUES (?, ?, ?, ?, ?)
""",
    orders
)

cursor.executemany(
    """
INSERT INTO order_items
(id, order_id, product_id, quantity, unit_price)
VALUES (?, ?, ?, ?, ?)
""",
    order_items
)

# ============================================================

# REVIEWS

# ============================================================

review_texts = {
    5: "Excellent product. Highly recommended.",
    4: "Good product and worth the money.",
    3: "Average product.",
    2: "Quality could be better.",
    1: "Very disappointing."
}

reviews = []

for review_id in range(1, 2501):

    product_id = random.randint(
        1,
        len(products)
    )

    customer_id = random.randint(
        1,
        500
    )

    rating = random.choices(
        [1, 2, 3, 4, 5],
        weights=[5, 5, 15, 30, 45],
        k=1
    )[0]

    review_date = (
        datetime.now()
        - timedelta(days=random.randint(0, 365))
    ).strftime("%Y-%m-%d")

    reviews.append(
        (
            review_id,
            product_id,
            customer_id,
            rating,
            review_texts[rating],
            review_date
        )
    )

cursor.executemany(
    """
INSERT INTO reviews
(id, product_id, customer_id, rating, review_text, review_date)
VALUES (?, ?, ?, ?, ?, ?)
""",
    reviews
)

conn.commit()

# ============================================================

# DATABASE SUMMARY

# ============================================================

print("\n" + "=" * 70)
print("DATABASE CREATED")
print("=" * 70)

tables = [
    "customers",
    "products",
    "orders",
    "order_items",
    "reviews"
]

for table in tables:

    result = pd.read_sql_query(
        "SELECT COUNT(*) AS total FROM " + table,
        conn
    )

    total = int(result.iloc[0]["total"])

    print(
        table.ljust(15),
        ":",
        total,
        "records"
    )

# ============================================================

# GEMINI

# ============================================================

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash"
)

print("\nGemini connected successfully.")

# ============================================================

# DATABASE SCHEMA

# ============================================================

SCHEMA = """

customers:
id, name, email, city, state, created_at

products:
id, name, category, price, cost, stock

orders:
id, customer_id, order_date, status, total_amount

order_items:
id, order_id, product_id, quantity, unit_price

reviews:
id, product_id, customer_id, rating, review_text, review_date

RELATIONSHIPS:

orders.customer_id = customers.id

order_items.order_id = orders.id

order_items.product_id = products.id

reviews.product_id = products.id

reviews.customer_id = customers.id

"""

# ============================================================

# BUSINESS RULES

# ============================================================

RULES = """

Revenue:

SUM(order_items.quantity * order_items.unit_price)

Profit:

SUM(
    order_items.quantity *
    (order_items.unit_price - products.cost)
)

Units sold:

SUM(order_items.quantity)

Average order value:

AVG(orders.total_amount)

Average rating:

AVG(reviews.rating)

Completed orders:

orders.status = 'completed'

Cancelled orders:

orders.status = 'cancelled'

Returned orders:

orders.status = 'returned'

Pending orders:

orders.status = 'pending'

For sales, revenue and profit questions,
normally use completed orders.

"""

# ============================================================

# GEMINI RESPONSE HANDLER

# ============================================================

def gemini_text(response):

    content = response.content

    if isinstance(content, str):
        return content.strip()

    if isinstance(content, list):

        output = []

        for item in content:

            if isinstance(item, str):
                output.append(
                    item
                )

            elif isinstance(item, dict):

                if "text" in item:
                    output.append(
                        str(item["text"])
                    )

        return "\n".join(output).strip()

    return str(content).strip()

# ============================================================

# CALL GEMINI

# ============================================================

def ask_gemini(prompt):

    response = llm.invoke(prompt)

    return gemini_text(response)

# ============================================================

# EXTRACT SQL

# ============================================================

def extract_sql(text):

    text = str(text).strip()

    match = re.search(
        r"```sql\s*(.*?)```",
        text,
        re.IGNORECASE | re.DOTALL
    )

    if match:
        return match.group(1).strip()

    match = re.search(
        r"```\s*(.*?)```",
        text,
        re.DOTALL
    )

    if match:
        return match.group(1).strip()

    text = re.sub(
        r"^SQL\s*:\s*",
        "",
        text,
        flags=re.IGNORECASE
    )

    return text.strip()

# ============================================================

# SQL SAFETY

# ============================================================

def safe_sql(sql):

    sql = sql.strip()

    upper_sql = sql.upper()

    if not (
        upper_sql.startswith("SELECT")
        or upper_sql.startswith("WITH")
    ):
        return False

    forbidden = [
        "INSERT",
        "UPDATE",
        "DELETE",
        "DROP",
        "ALTER",
        "CREATE",
        "TRUNCATE",
        "REPLACE",
        "ATTACH",
        "DETACH",
        "VACUUM",
        "PRAGMA"
    ]

    for word in forbidden:

        pattern = r"\b" + word + r"\b"

        if re.search(pattern, upper_sql):
            return False

    return True

# ============================================================

# GENERATE SQL

# ============================================================

def generate_sql(question):

    prompt = f"""
You are an expert e-commerce SQL analyst.

Convert the user's question into ONE valid SQLite SQL query.

SCHEMA:

{SCHEMA}

BUSINESS RULES:

{RULES}

USER QUESTION:

{question}

RULES:

1. Return ONLY SQL.
2. Do not use markdown.
3. Do not explain anything.
4. Only use tables and columns in the schema.
5. Use correct JOIN conditions.
6. Use GROUP BY when required.
7. Use ORDER BY for ranking questions.
8. Use LIMIT for top/bottom questions.
9. Use SQLite syntax.
10. Only SELECT or WITH queries are allowed.
11. Never modify data.

SQL:
"""

    result = ask_gemini(prompt)

    sql = extract_sql(result)

    return sql

# ============================================================

# REPAIR SQL

# ============================================================

def repair_sql(
    question,
    failed_sql,
    error_message
):

    prompt = f"""
You are an expert SQLite SQL debugger.

Fix the SQL query.

SCHEMA:

{SCHEMA}

BUSINESS RULES:

{RULES}

QUESTION:

{question}

FAILED SQL:

{failed_sql}

ERROR:

{error_message}

Return ONLY the corrected SQL query.

Rules:

* SELECT or WITH only.
* No INSERT.
* No UPDATE.
* No DELETE.
* No DROP.
* No ALTER.
* No CREATE.
* No markdown.
* No explanation.

CORRECTED SQL:
"""

    result = ask_gemini(prompt)

    return extract_sql(result)

# ============================================================

# EXECUTE QUESTION

# ============================================================

def execute_question(question):

    sql = generate_sql(question)

    last_error = ""

    for attempt in range(3):

        if not safe_sql(sql):

            last_error = "Unsafe SQL query."

            sql = repair_sql(
                question,
                sql,
                last_error
            )

            continue

        try:

            dataframe = pd.read_sql_query(
                sql,
                conn
            )

            return sql, dataframe

        except Exception as error:

            last_error = str(error)

            sql = repair_sql(
                question,
                sql,
                last_error
            )

    raise Exception(
        "SQL generation failed after 3 attempts.\n"
        + last_error
    )

# ============================================================

# ANALYZE RESULT

# ============================================================
def analyze_result(question, sql, dataframe):

    if dataframe.empty:
        return "😂 Oops! The database found nothing. It's hiding from us! 😜"

    result_text = dataframe.head(50).to_string(index=False)

    prompt = f"""
    ```

    You are a very funny, mischievous, energetic cartoon-style
    e-commerce AI analyst.

    USER QUESTION:
    {question}

    SQL:
    {sql}

    DATABASE RESULT:
    {result_text}

    Answer the user's question using ONLY the database result.

    PERSONALITY:

    * Be extremely funny and playful.
    * Use short jokes.
    * Use dramatic reactions.
    * Use emojis naturally.
    * Sound like a mischievous cartoon character.
    * Keep the humor family-friendly.

    Use expressions such as:
    WHOA! 🤯
    OH BOY! 😂
    WAIT... WHAT?! 😱
    AHAHA! 🤣
    HEHEHE! 😜
    OHOHO! 🤑
    YIKES! 😵

    ACCURACY RULES:

    * Always answer the question directly.
    * Never invent numbers.
    * Never change numbers.
    * Never exaggerate numbers.
    * Use only information from the database result.
    * Use ₹ for money.
    * Show rankings clearly.
    * Keep the business analysis accurate.
    * Humor must never replace the answer.
    * Add 1 or 2 short jokes after the factual answer.

    RESPONSE FORMAT:

    🎯 ANSWER
    Give the direct answer.

    📊 DETAILS
    Explain the important numbers.

    😂 CARTOON COMMENT
    Add a short funny comment.

    Example:

    WHOA! 🤯 We found the sales superstar!

    🏆 Laptop Pro 14 generated ₹18,42,350 in revenue.

    📊 It is the highest-revenue product in the result.

    😂 CARTOON COMMENT:
    That laptop isn't processing data anymore...
    it's processing MONEY! 💰🤣

    Now answer the user's question.

    FINAL ANSWER:
    """

    try:
        response = llm.invoke(prompt)
        return gemini_text(response)

    except Exception as error:
        return "😂 Oops! My cartoon brain crashed! Error: " + str(error)



# ============================================================

# MAIN ANALYST

# ============================================================

def ask_analyst(question):

    question = str(question).strip()

    if not question:

        return "Please enter a question."

    try:

        print("\n🔎 Generating SQL...")

        sql, dataframe = execute_question(
            question
        )

        print("\n🧾 SQL")
        print("-" * 70)

        print(sql)

        print("\n📊 DATA")
        print("-" * 70)

        if dataframe.empty:

            print("No results.")

        else:

            print(
                dataframe.head(20).to_string(
                    index=False
                )
            )

        print("\n🧠 ANALYSIS")
        print("-" * 70)

        answer = analyze_result(
            question,
            sql,
            dataframe
        )

        return answer

    except Exception as error:

        return (
            "❌ Error:\n\n"
            + str(error)
        )

# from IPython.display import Javascript, display
# import html

def speak_text(text):
    text = str(text)

    # Remove some formatting that doesn't sound useful when spoken
    text = re.sub(r'```.*?```', '', text, flags=re.DOTALL)
    text = re.sub(r'[*#`]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()

    if not text:
        return

    try:
        tts = gTTS(text=text, lang='en', slow=False)
        tts.save("speech.mp3")
        display(Audio("speech.mp3", autoplay=True))
    except Exception as e:
        print(f"Text-to-speech error: {e}")

# ============================================================

# TEST QUESTIONS

# ============================================================

questions = [

    "Which products generated the most revenue?",

    "What is the total profit?",

    "Which city has the most customers?",

    "Which category generates the most revenue?",

    "What is the cancellation rate?",

    "What is the average order value?",

    "Which products have the worst ratings?",

    "Which products have less than 20 units in stock?",

    "Which category is the most profitable?",

    "Show monthly revenue.",

    "Compare Electronics and Fashion revenue.",

    "Which state generates the most revenue?",

    "Show the top 10 customers by spending.",

    "Which products have the highest profit margin?",

    "How many orders were returned?",

    "How many completed orders do we have?",

    "Which product sold the most units?",

    "What is the average product rating?",

    "Which city generated the most revenue?",

    "Show revenue by category."

]

# ============================================================

# DISPLAY

# ============================================================

print("\n" + "=" * 70)
print("🤖 E-COMMERCE AI DATA ANALYST READY")
print("=" * 70)

print("\nExample questions:")

for number, question in enumerate(
    questions,
    start=1
):

    print(
        number,
        "-",
        question
    )

# ============================================================

# AUTOMATIC TEST

# ============================================================

print("\n" + "=" * 70)
print("🧪 AUTOMATIC TEST")
print("=" * 70)

test_question = (
    "Which products generated the most revenue?"
)

print("\nQuestion:")
print(test_question)

test_answer = ask_analyst(
    test_question
)

print("\nFINAL ANSWER")
print("-" * 70)
print(test_answer)
speak_text(test_answer)

# ============================================================

# INTERACTIVE CHAT

# ============================================================

print("\n" + "=" * 70)
print("💬 INTERACTIVE CHAT")
print("=" * 70)

print("""
You can now ask questions about the database.

Examples:

Which products generated the most revenue?

What is the total profit?

Which city has the most customers?

Which category generates the most revenue?

What is the cancellation rate?

What is the average order value?

Which products have the worst ratings?

Which products have less than 20 units in stock?

Which category is the most profitable?

Show monthly revenue.

Compare Electronics and Fashion revenue.

Which state generates the most revenue?

Show the top 10 customers by spending.

Which products have the highest profit margin?

How many orders were returned?

Type EXIT to stop.
""")

while True:

    question = input("\n🧑 You: ").strip()

    if question.lower() in ["exit", "quit", "q"]:
        print("\n👋 Goodbye!")
        break

    if not question:
        continue

    print("\n" + "=" * 70)
    print("🤖 AI ANALYST")
    print("=" * 70)

    answer = ask_analyst(question)

    print("\nFINAL ANSWER")
    print("-" * 70)
    print(answer)
    print("=" * 70)

    speak_text(answer)
````
